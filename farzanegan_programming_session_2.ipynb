### چرا رشته مهم است؟
تقریباً هر برنامه‌ای با متن کار می‌کند: نام کاربر، پیام خوش‌آمدگویی، مسیر فایل، شماره تلفن و ...  
رشته در پایتون یک **شیء** است که خودش طولش را می‌داند و بعد از ساخته شدن تغییر نمی‌کند (به این ویژگی می‌گویند **تغییرناپذیر** یا immutable). یعنی اگر بخواهی یک حرف وسطش را عوض کنی، باید یک رشتهٔ جدید بسازی.

### ۱.۱ ساختن و چاپ رشته + دسترسی به کاراکترها

**مفهوم ایندکس (Index):**  
هر کاراکتر در رشته یک شماره دارد که از **صفر** شروع می‌شود.  
- اولین حرف → ایندکس `0`  
- دومین حرف → ایندکس `1`  
- و الی آخر  

اگر ایندکس خارج از محدوده باشد، پایتون خطای `IndexError` می‌دهد.
</div>


````python
s = "HI!"
print(s)                 # چاپ کل رشته

# دسترسی به تک‌تک کاراکترها با ایندکس
print(s[0])              # H   ← اولین کاراکتر
print(s[1])              # I
print(s[2])              # !

# طول رشته
print("طول رشته:", len(s))
````

<div class="rtl-box" dir="rtl" style="direction:rtl; text-align:right; font-family:'Vazirmatn',Tahoma,'Segoe UI',Arial,sans-serif; line-height:1.95;">
> **نکتهٔ خیلی مهم:** ایندکس از **۰** شروع می‌شود. این یک قرارداد رایج در برنامه‌نویسی است؛ عادت کن به آن.

### ۱.۲ کاراکتر و کد عددی

کامپیوتر در نهایت همه چیز را به صورت عدد ذخیره می‌کند.  
هر حرف یک شمارهٔ مخصوص دارد (مثلاً حرف `H` برابر با ۷۲ است).

در پایتون:
- `ord(کاراکتر)` → عدد مربوط به آن کاراکتر را برمی‌گرداند  
- `chr(عدد)` → از عدد به کاراکتر برمی‌گردد
</div>


````python
c = '#'
print("کاراکتر:", c)
print("کد عددی:", ord(c))

s = "HI!"
print("\nکاراکتر و کد عددی هر حرف:")
for ch in s:
    print(ch, ord(ch))

# از عدد به کاراکتر برگردیم
print("\nchr(72) =", chr(97))      # باید H چاپ شود
````

<div class="rtl-box" dir="rtl" style="direction:rtl; text-align:right; font-family:'Vazirmatn',Tahoma,'Segoe UI',Arial,sans-serif; line-height:1.95;">
### ۱.۳ طول رشته

برای دانستن تعداد کاراکترهای یک رشته، کافی است از تابع آمادهٔ `len()` استفاده کنی.  
گاهی برای درک بهتر، می‌توانیم خودمان هم با یک حلقه طول را بشماریم؛ اما در کد واقعی همیشه از `len()` استفاده کن.
</div>


````python
# روش دستی — فقط برای یادگیری مفهوم
def string_length(text):
    """طول رشته را دستی می‌شمارد."""
    n = 0
    while n < len(text):
        n += 1
    return n

# روش استاندارد و پیشنهادی
name = input("اسمت را بنویس: ")
print("طول با تابع دستی:", string_length(name))
# print("طول با len():", len(name))
````

<div class="rtl-box" dir="rtl" style="direction:rtl; text-align:right; font-family:'Vazirmatn',Tahoma,'Segoe UI',Arial,sans-serif; line-height:1.95;">
> **درس مهندسی:** وقتی زبان یک ابزار آماده، امن و سریع دارد (`len`)، از آن استفاده کن.  
> پیاده‌سازی دستی فقط برای درک عمیق مفهوم است، نه برای کد واقعی پروژه.

### ۱.۴ چاپ کاراکتر به کاراکتر

گاهی می‌خواهیم تک‌تک حروف یک رشته را ببینیم یا روی آن‌ها عملیات انجام دهیم.  
دو روش رایج وجود دارد:
1. با ایندکس و `range(len(s))`  
2. مستقیم روی کاراکترها (`for ch in s`) — ساده‌تر و خواناتر
</div>


````python
s = input("ورودی: ")
print("خروجی: ", end="")

# روش ۱: با range و len
for i in range(len(s)):
    print(s[i], end="")
    print("")
print()   # خط جدید

# روش ۲: مستقیم روی کاراکترها (پیشنهادی)
print("خروجی ساده‌تر: ", end="")
for ch in s:
    print(ch, end="")
print()
````

<div class="rtl-box" dir="rtl" style="direction:rtl; text-align:right; font-family:'Vazirmatn',Tahoma,'Segoe UI',Arial,sans-serif; line-height:1.95;">
### ۱.۵ تبدیل به حروف بزرگ

گاهی لازم است متن را به حروف بزرگ تبدیل کنیم (مثلاً برای مقایسهٔ یکسان).  
دو راه داریم:
1. روش دستی — برای درک اینکه کامپیوتر چطور کار می‌کند  
2. متد آمادهٔ `.upper()` که امن‌تر و خواناتر است

**توجه:** رشته تغییرناپذیر است؛ `.upper()` یک رشتهٔ **جدید** برمی‌گرداند و رشتهٔ اصلی را عوض نمی‌کند.
</div>


````python
s = input("قبل: ")
print("بعد (روش دستی):  ", end="")

# روش دستی
for ch in s:
    if 'a' <= ch <= 'z':
        print(chr(ord(ch) - 32), end="")
    else:
        print(ch, end="")
print()

# روش استاندارد (پیشنهادی برای همهٔ پروژه‌ها)
print("با متد upper():", s.upper())
print("با متد title():", s.title())   # حرف اول هر کلمه بزرگ می‌شود
````

> **سؤال سقراطی:** چرا رشته در پایتون بعد از ساخته شدن تغییر نمی‌کند؟ اگر بخواهیم یک حرف وسط رشته را عوض کنیم چه باید بکنیم؟

  <h2 style="margin: 0; font-size: 24px;">۲. لیست (List) — جعبهٔ چندخانه‌ای انعطاف‌پذیر</h2>

### چرا لیست؟
وقتی چند مقدار مرتبط داریم (مثلاً سه نمره، چند نام درس، چند شماره تلفن)، به‌جای ساختن متغیرهای جداگانه، آن‌ها را داخل یک **لیست** می‌ریزیم.  
لیست می‌تواند بزرگ یا کوچک شود و انواع مختلف داده را نگه دارد (عدد، رشته، حتی لیست دیگر).

لیست هم مثل رشته ایندکس از صفر دارد:
```
scores = [72, 73, 33]
           ↑   ↑   ↑
          [0] [1] [2]
```

### ۲.۱ میانگین نمرات — از متغیر جدا تا لیست
</div>


````python
# روش ساده با متغیرهای جدا — فقط برای تعداد کم خوب است
score1 = 72
score2 = 73
score3 = 33
print("میانگین (متغیر جدا):", (score1 + score2 + score3) / 3)

# با لیست — بهتر و مقیاس‌پذیرتر
scores = [72, 73, 33]
print("میانگین با لیست:", sum(scores) / len(scores))

# گرفتن نمرات از کاربر و ریختن در لیست
scores = []
for i in range(3):
    n = float(input(f"نمره {i+1}: "))
    scores.append(n)          # append یعنی اضافه کردن به انتهای لیست

print("نمرات:", scores)
print("میانگین:", sum(scores) / len(scores))
````

<div class="rtl-box" dir="rtl" style="direction:rtl; text-align:right; font-family:'Vazirmatn',Tahoma,'Segoe UI',Arial,sans-serif; line-height:1.95;">
### ۲.۲ نوشتن تابع برای میانگین

**چرا تابع؟**  
- کد تکراری را یک‌بار می‌نویسیم و بارها استفاده می‌کنیم.  
- نام تابع خودش توضیح می‌دهد چه کاری انجام می‌دهد (`average`).  
- اگر بعداً بخواهیم فرمول را عوض کنیم، فقط یک جا تغییر می‌دهیم.

این همان مفهوم **Abstraction** است: جزئیات غیرضروری را پنهان کن تا کد خواناتر شود.
</div>


````python
def average(numbers):
    """میانگین یک لیست از اعداد را برمی‌گرداند."""
    if len(numbers) == 0:
        return 0.0          # جلوگیری از تقسیم بر صفر
    return sum(numbers) / len(numbers)

# تست تابع
scores = [2,2,2,5.3]


print(f"میانگین: {average(scores):.4f}")   # .2f یعنی دو رقم اعشار
````

<div class="rtl-box" dir="rtl" style="direction:rtl; text-align:right; font-family:'Vazirmatn',Tahoma,'Segoe UI',Arial,sans-serif; line-height:1.95;">
### ۲.۳ لیستی از رشته‌ها

لیست فقط برای عدد نیست. می‌توانیم لیستی از رشته‌ها هم داشته باشیم (مثلاً لیست دروس یا نام‌ها).
</div>


````python
words = ["HI!", "BYE!"]

for w in words:
    print(w)

# یا با ایندکس (اگر شمارهٔ ردیف هم لازم باشد)
for i in range(len(words)):
    print(i, words[i])
````

  <h2 style="margin: 0; font-size: 24px;">۳. جستجوی خطی (Linear Search)</h2>
</div>

### مسئله چیست؟
یک لیست از مقادیر داریم و می‌خواهیم بفهمیم آیا مقدار خاصی داخل آن هست یا نه.  
خروجی معمولاً یک جواب بله/خیر است.

### ساده‌ترین الگوریتم: جستجوی خطی
از اول لیست شروع می‌کنیم و یکی‌یکی نگاه می‌کنیم تا یا مقدار را پیدا کنیم یا به آخر لیست برسیم.

**به زبان ساده:**
```
برای هر عنصر از اول تا آخر:
    اگر عنصر برابر با هدف بود:
        بگو پیدا شد و تمام کن
اگر به آخر رسیدی و پیدا نکردی:
    بگو پیدا نشد
```

در بدترین حالت باید همهٔ عناصر را ببینیم. به این می‌گویند پیچیدگی **O(n)**؛ یعنی زمان تقریباً متناسب با تعداد عناصر است.
</div>


````python
numbers = [4, 6, 8, 2, 7, 5, 0]
lenNum = len(numbers)

target = int(input("addad morede nazar ra begid :")) 

found = False

place = 0

for i in range(lenNum):
    if numbers[i] == target:
        found = True
        place = i
        break          # پیدا شد؛ دیگر لازم نیست ادامه بدهیم

if found:
    print("پیدا شد injast :: " , place)
else:
    print("پیدا نشد")
````

### دو راه برای فکر کردن به یک مسئله

**تکرار (Iteration):** با حلقه (`for` یا `while`) قدم‌به‌قدم جلو می‌رویم.  
**بازگشت (Recursion):** مسئله را به نسخهٔ کوچک‌تر خودش تبدیل می‌کنیم تا به حالت پایه برسیم.

هر دو روش می‌توانند یک هرم از `#` بسازند. تفاوت در **نحوهٔ فکر کردن** است، نه فقط در نوشتن کد.

### قوانین طلایی بازگشت
1. **حالت پایه:** وقتی دیگر نباید ادامه بدهیم (اینجا وقتی ارتفاع صفر یا منفی است).  
2. **گام بازگشتی:** مسئله را کوچک‌تر کن و دوباره همان تابع را صدا بزن.  

اگر حالت پایه را فراموش کنی، برنامه تا ابد خودش را صدا می‌زند و خطای `RecursionError` می‌گیری.
</div>


````python
def draw_iterative(n):
    """هرم را با حلقه رسم می‌کند."""
    for i in range(n):
        print("#" * (i + 1))

height = int(input("ارتفاع هرم (روش تکراری): "))
draw_iterative(height)
````

````python
def draw_recursive(n):
    """هرم را با بازگشت رسم می‌کند."""
    # ۱. حالت پایه
    if n <= 0:
        return
    
    # ۲. اول هرم کوچک‌تر را بکش
    draw_recursive(n - 1)
    
    # ۳. بعد ردیف فعلی را چاپ کن
    print("#" * n)

height = int(input("ارتفاع هرم (روش بازگشتی): "))
draw_recursive(height)
````

  <h2 style="margin: 0; font-size: 24px;">۶. اسکلت پروژه — نسخه ۰.۲ دستیار مطالعه</h2>
</div>

در این نسخه از **همهٔ مفاهیم امروز** استفاده می‌کنیم:
- لیست دیکشنری برای ذخیره دروس
- جستجوی خطی برای پیدا کردن درس
- تابع برای کارهای تکراری
- اعتبارسنجی ورودی (جلوگیری از دادهٔ نامعتبر)

کد را با دقت بخوان، بعد اجرا کن و با آن بازی کن.

````python
def print_menu():
    print("\n===== دستیار هوشمند مطالعه (نسخه ۰.۲) =====")
    print("1. افزودن درس جدید")
    print("2. نمایش همه دروس")
    print("3. جستجوی درس")
    print("4. تعداد کل دروس")
    print("5. خروج")

def add_subject(study_list):
    name = input("نام درس: ").strip()
    if not name:
        print("نام نمی‌تواند خالی باشد.")
        return
    
    priority = input("اولویت (بالا / متوسط / پایین): ").strip()
    valid = ["بالا", "متوسط", "پایین"]
    if priority not in valid:
        print("اولویت نامعتبر است. یکی از این‌ها را بنویس:", "، ".join(valid))
        return
    
    study_list.append({"name": name, "priority": priority})
    print("✓ درس اضافه شد.")

def show_subjects(study_list):
    if len(study_list) == 0:
        print("هنوز درسی اضافه نکرده‌ای.")
        return
    print("\nلیست دروس تو:")
    for i, item in enumerate(study_list, start=1):
        print(f"{i}. {item['name']} | اولویت: {item['priority']}")

def search_subject(study_list):
    target = input("نام درس برای جستجو: ").strip()
    found = False
    for item in study_list:
        if item["name"] == target:
            print(f"پیدا شد → اولویت: {item['priority']}")
            found = True
            break
    if not found:
        print("درس پیدا نشد.")

# --- برنامه اصلی ---
study_list = []

while True:
    print_menu()
    choice = input("گزینه مورد نظر: ").strip()
    
    if choice == "1":
        add_subject(study_list)
    elif choice == "2":
        show_subjects(study_list)
    elif choice == "3":
        search_subject(study_list)
    elif choice == "4":
        print("تعداد کل دروس:", len(study_list))
    elif choice == "5":
        print("برنامه بسته شد. جلسه بعد ادامه می‌دهیم...")
        break
    else:
        print("گزینه اشتباه است.")
````
