### چهار ستون تفکر مهندسی

| مفهوم | معنی برای ما |
|--------|-------------|
| **کدنویسی** | ابزار است، نه هدف |
| **حل مسئله** | هسته اصلی کار مهندس |
| **Trade-off** | هر تصمیمی چیزی را بهتر و چیزی را بدتر می‌کند |
| **انسان‌ها** | کاربر، تیم، ذی‌نفعان — نرم‌افزار برای آدم‌هاست |

### ایده مرکزی
> مهندسی نرم‌افزار = تبدیل **مسئله واقعی** به **سیستم قابل اتکا، مقیاس‌پذیر و قابل نگهداری**

سه مرحله‌ای که امروز شروع می‌کنیم:

1. **درک مسئله** → Requirements، کاربران، محدودیت‌ها  
2. **طراحی** → Abstraction، ساختار، Trade-off  
3. **ساخت و اطمینان** → پیاده‌سازی، تست، بازخورد



<div style="background: #0F172A; color: #F8FAFC; padding: 24px 28px; border-radius: 12px; border-right: 5px solid #F59E0B; margin: 20px 0;">
  <h2 style="margin: 0; font-size: 24px;">از مشکل روزمره تا مسئله نرم‌افزاری</h2>
</div>


1. یک مشکل واقعی از زندگی دانش‌آموزی انتخاب کنید  
   *(برنامه‌ریزی مطالعه، پیدا کردن گروه پروژه، مدیریت زمان آزمون‌ها، ...)*

2. کاربران اصلی را مشخص کنید. چه نیازهایی دارند؟ چه محدودیت‌هایی؟

3. مسئله را به زبان نرم‌افزاری بازنویسی کنید:  
   - **ورودی** چیست؟  
   - **خروجی** چیست؟  
   - **رفتار مورد انتظار** چیست؟

4. حداقل **۲ Trade-off** بنویسید  
   *(مثلاً: سادگی در برابر قدرت · سرعت در برابر دقت)*

### سؤال‌های راهنما برای گروه

| سؤال | نکته |
|------|------|
| کاربر دقیقاً چه می‌خواهد؟ | نه آنچه که *فکر می‌کنید* می‌خواهد |
| اگر سیستم اشتباه کند چه می‌شود؟ | هزینه خطا چیست؟ |
| چه چیزی را باید پنهان کنیم؟ | Abstraction = جزئیات غیرضروری را پنهان کن |
| چه چیزی را باید قابل تغییر نگه داریم؟ | نرم‌افزار تغییر می‌کند |

### ۱. برنامه با کاربر حرف می‌زند

`print` برای نمایش پیام · `input` برای دریافت داده از کاربر


```python
print("سلام! به دستیار مطالعه خوش آمدی.")
name = input("اسمت چیه؟ ")
print("خوش اومدی، " + name + "!")
```


### ۲. متغیر و محاسبه

داده را ذخیره می‌کنیم و روی آن عملیات انجام می‌دهیم.  
`float()` ورودی متنی را به عدد اعشاری تبدیل می‌کند.


```python
score1 = float(input("نمره اول را وارد کن: "))
score2 = float(input("نمره دوم را وارد کن: "))
average = (score1 + score2) / 2
print("میانگین تو:", average)
```


> **اشتباه رایج:** فراموش کردن `float()` یا `int()` → خطای TypeError  
> از خودتان بپرسید: چرا نمی‌توانیم مستقیم `"18" + "19"` را تقسیم کنیم؟



### ۳. شرط — برنامه تصمیم می‌گیرد

`if` / `elif` / `else` مسیر اجرا را بر اساس شرط تغییر می‌دهند.


```python
hours = float(input("امروز چند ساعت مطالعه کردی؟ "))

if hours >= 4:
    print("عالی! خیلی خوب پیش رفتی.")
elif hours >= 2:
    print("خوبه، ولی می‌تونی بهتر هم باشی.")
else:
    print("امروز کم کار کردی. فردا جبران کن!")
```


> **تمرین سریع:** شرط را طوری تغییر دهید که برای ساعت‌های منفی هم پیام مناسب بدهد.  
> **سؤال سقراطی:** ترتیب شرط‌ها مهم است؟ اگر `>= 2` را اول بنویسیم چه می‌شود؟



### ۴. حلقه for — تکرار هوشمند

به‌جای نوشتن کد تکراری، از حلقه استفاده می‌کنیم.


```python
subjects = ["ریاضی", "فیزیک", "شیمی", "ادبیات"]
scores = []

print("دروس امروز:")
for i, subject in enumerate(subjects, start=1):
    score = float(input("نمره   " + subject + "را وارد کن:  "))
    scores.append(score)
    print(i, "-", subject , "-",  score) 


average = sum(scores) / len(subjects)
print("معدل تو:", average)
```

```python
numbers = int(input("تعداد دروس: "))
scores = []
sumScores = 0

for number in range(numbers):
    score = float(input("نمره را وارد کن: "))
    scores.append(score)
    sumScores += score

average = float(sumScores / numbers)
print("میانگین تو:", average)
```


> **مفهوم کلیدی:** `list` (لیست) یکی از مهم‌ترین ساختارهای داده است.  
> در جلسه ۲ عمیق‌تر با آن کار می‌کنیم.



### ۵. حلقه while + منوی تعاملی

تا وقتی کاربر نخواهد، برنامه ادامه می‌دهد.


```python
while True:
    print("\n--- منوی دستیار مطالعه ---")
    print("1. افزودن درس")
    print("2. مشاهده لیست دروس")
    print("3. خروج")
    
    choice = input("انتخاب تو: ")
    
    if choice == "1":
        subject = input("نام درس را بنویس: ")
        print("درس", subject, "اضافه شد.")
    elif choice == "2":
        print("فعلاً لیست خالی است (در ادامه کامل می‌شود).")
    elif choice == "3":
        print("خداحافظ! موفق باشی.")
        break
    else:
        print("گزینه نامعتبر است. دوباره امتحان کن.")
```


> **نکته مهم:** `break` حلقه را متوقف می‌کند.  
> بدون `break` برنامه تا ابد اجرا می‌شود (Infinite Loop).

### آنچه برای جلسه بعد باید بسازیم
- افزودن درس با اولویت
- نمایش لیست دروس
- شمارش تعداد دروس
- منوی تعاملی کامل


```python
print("===== دستیار هوشمند مطالعه (نسخه ۰.۱) =====")

# ذخیره دروس در یک لیست از دیکشنری‌ها
study_list = []

while True:
    print("\nمنو:")
    print("1. افزودن درس جدید")
    print("2. نمایش همه دروس")
    print("3. محاسبه تعداد کل دروس")
    print("4. خروج")
    
    choice = input("گزینه مورد نظر: ")
    
    if choice == "1":
        subject = input("نام درس: ")
        priority = input("اولویت (بالا / متوسط / پایین): ")
        study_list.append({"name": subject, "priority": priority})
        print("✓ درس اضافه شد.")
        
    elif choice == "2":
        if len(study_list) == 0:
            print("هنوز درسی اضافه نکرده‌ای.")
        else:
            print("\nلیست دروس تو:")
            for i, item in enumerate(study_list, start=1):
                print(i, "-", item["name"], "| اولویت:", item["priority"])
                
    elif choice == "3":
        print("تعداد کل دروس:", len(study_list))
        
    elif choice == "4":
        print("برنامه بسته شد. جلسه بعد ادامه می‌دهیم...")
        break
        
    else:
        print("گزینه اشتباه است.")
```

### 🔥 Challenge برای دانش‌آموزان قوی‌تر

تابع زیر را فعال و کامل کنید تا ورودی نامعتبر قبول نشود:


```python
def get_valid_hours():
    """تا وقتی کاربر عدد معتبر وارد نکند، دوباره میپرسد."""
    while True:
        try:
            hours = float(input("چند ساعت مطالعه کردی؟ "))
            if hours < 0:
                print("ساعت نمیتواند منفی باشد.")
                continue
            return hours
        except ValueError:
            print("لطفا یک عدد وارد کن.")

# تست تابع
h = get_valid_hours()
print("ثبت شد:", h, "ساعت")
```

### اشتباهات رایج دانش‌آموزان
- فراموش کردن `float()` / `int()` روی `input`
- تورفتگی (indentation) اشتباه بعد از `if` و `for`
- استفاده از `=` به‌جای `==` در شرط
- نام‌گذاری ضعیف متغیر (`s` به‌جای `subject`)
- ترس از خطا — خطا دوست ماست، پیامش را بخوان
