<div class="dark"><h2 style="margin-top:0;">۱. الگوریتم چیست؟</h2></div>

### تعریف ساده

**الگوریتم = یک دستورالعمل دقیق و مرحله‌به‌مرحله برای تبدیل ورودی به خروجی.**

کد، فقط یکی از راه‌های بیان الگوریتم است. قبل از Python، باید بدانیم «چه کاری» می‌خواهیم انجام دهیم.

مثال روزمره: پیدا کردن بزرگ‌ترین عدد در یک برگه.

- ورودی: چند عدد
- هدف: بزرگ‌ترین عدد
- ایده:
  1. عدد اول را موقتاً بزرگ‌ترین فرض کن.
  2. بقیه را یکی‌یکی ببین.
  3. اگر عدد جدید بزرگ‌تر بود، جواب موقت را عوض کن.
  4. در پایان، جواب موقت همان جواب اصلی است.

این یک الگوریتم است؛ حتی اگر هنوز هیچ کدی ننوشته باشیم.

### یک الگوریتم خوب حداقل این ویژگی‌ها را دارد

- **دقیق** است: هر مرحله معلوم است.
- **پایان‌پذیر** است: تا ابد ادامه پیدا نمی‌کند.
- **درست** است: برای ورودی‌های معتبر جواب درست می‌دهد.
- **قابل اجرا** است: مراحلش واقعاً قابل انجام‌اند.
- بهتر است **کارا** هم باشد: زمان و حافظهٔ بیهوده مصرف نکند.
</div>

````python
numbers = [8, 3, 12, 5, 9]

best = numbers[0]
for x in numbers[1:]:
    if x > best:
        best = x

print("بزرگ‌ترین عدد:", best)
````

<div class="dark"><h2 style="margin-top:0;">۲. صورت مسئله را به چهار بخش بشکن</h2></div>

وقتی یک سؤال می‌بینی، هنوز سراغ کد نرو. این چهار سؤال را پاسخ بده:

1. **ورودی چیست؟**
2. **خروجی چیست؟**
3. **محدودیت‌ها چیست؟**
4. **رابطهٔ ورودی و خروجی چیست؟**

### مثال

> یک لیست از نمره‌ها داده شده است. تعداد نمره‌های بزرگ‌تر یا مساوی ۱۰ را پیدا کنید.

- ورودی: لیستی از عددها
- خروجی: یک عدد صحیح
- محدودیت مهم: تعداد نمره‌ها می‌تواند کوچک یا خیلی بزرگ باشد.
- رابطه: هر عضو را بررسی می‌کنیم؛ اگر شرط را داشت، شمارنده یک واحد زیاد می‌شود.

### شبه‌کد (Pseudo-code)

شبه‌کد زبان برنامه‌نویسی نیست. برای روشن کردن فکر است:

```text
count = 0
برای هر نمره:
    اگر نمره >= 10:
        count را یکی زیاد کن
count را اعلام کن
```

بعد از روشن شدن ایده، نوشتن Python ساده می‌شود.
</div>

````python
scores = [8, 17, 10, 6, 15, 9, 20]

count = 0
for score in scores:
    if score >= 10:
        count += 1

print("تعداد قبولی‌ها:", count)
````

<div class="dark"><h2 style="margin-top:0;">۴. لیست/آرایه — اولین مدل ذهنی ساختمان داده</h2></div>

در منابع الگوریتم معمولاً زیاد واژهٔ **Array** را می‌بینی. در Python برای شروع، `list` نزدیک‌ترین ابزار روزمرهٔ ماست.

```text
index:   0    1    2    3    4
value:  12    5    9    3    7
```

- هر خانه یک **ایندکس** دارد.
- ایندکس از صفر شروع می‌شود.
- با `a[i]` مستقیماً به خانهٔ `i` می‌رویم.
- ترتیب اعضا مهم است.

در جلسهٔ سوم تقریباً همه‌جا با «دنباله»، «آرایه»، «حالت‌ها» و «ایندکس‌ها» سروکار داریم؛ پس این تصویر ذهنی بسیار مهم است.
</div>

````python
a = [12, 5, 9, 3, 7]

print("a[0] =", a[0])
print("a[3] =", a[3])

a[1] = 100
print("بعد از تغییر:", a)
````

### ۴.۱ عملیات مهم روی لیست

```python
a.append(x)     # افزودن به انتها
a.pop()         # حذف از انتها
a[i]            # دسترسی با ایندکس
x in a          # بررسی وجود عضو
len(a)          # تعداد اعضا
```

نکتهٔ مهم برای بعد:

- دسترسی به `a[i]` را می‌توانیم تقریباً «مستقیم» انجام دهیم.
- اما برای `x in a` در یک لیست نامرتب، ممکن است مجبور شویم اعضا را یکی‌یکی ببینیم.

همین تفاوت ما را به مفهوم **الگوریتم جست‌وجو** می‌رساند.

<div class="rtl-box">
<div class="dark"><h2 style="margin-top:0;">۵. جست‌وجوی خطی (Linear Search)</h2></div>

فرض کن کارت‌هایی به ترتیب روی میز هستند و می‌خواهی کارت `9` را پیدا کنی:

```text
[4, 6, 8, 2, 9, 5, 1]
```

ساده‌ترین راه:

1. از اول شروع کن.
2. هر عضو را با هدف مقایسه کن.
3. اگر پیدا شد، متوقف شو.
4. اگر به آخر رسیدی، وجود ندارد.

این همان **Linear Search** است.
</div>

````python
def linear_search(a, target):
    for i in range(len(a)):
        if a[i] == target:
            return i
    return -1

numbers = [4, 6, 8, 2, 9, 5, 1]
print("ایندکس 9:", linear_search(numbers, 9))
print("ایندکس 100:", linear_search(numbers, 100))
````

<div class="rtl-box">
### ۵.۱ نسخهٔ آموزشی با نمایش گام‌ها

این بار می‌خواهیم «هزینهٔ کار» را ببینیم.
</div>

````python
def linear_search_trace(a, target):
    comparisons = 0
    for i, x in enumerate(a):
        comparisons += 1
        print(f"گام {comparisons}: a[{i}]={x} را با {target} مقایسه می‌کنیم")
        if x == target:
            print("✅ پیدا شد")
            return i, comparisons
    print("❌ پیدا نشد")
    return -1, comparisons

linear_search_trace([4, 6, 8, 2, 9, 5, 1], 9)
````

### ۵.۲ بهترین و بدترین حالت

در لیست ۱۰۰۰ عضوی:

- اگر هدف عضو اول باشد → فقط ۱ مقایسه
- اگر عضو آخر باشد → ۱۰۰۰ مقایسه
- اگر اصلاً نباشد → ۱۰۰۰ مقایسه

برای طراحی الگوریتم معمولاً می‌پرسیم:

> وقتی اندازهٔ ورودی بزرگ شود، تعداد کارها چگونه رشد می‌کند؟

### Big-O چه چیزی را می‌گوید؟

Big-O به‌طور ساده می‌گوید:

> با بزرگ شدن اندازهٔ ورودی، مقدار کار الگوریتم با چه سرعتی رشد می‌کند؟

اگر `n` تعداد اعضای ورودی باشد:

- `O(1)` → مقدار کار تقریباً مستقل از `n`
- `O(log n)` → با هر گام بخش بزرگی از مسئله حذف می‌شود
- `O(n)` → تقریباً به ازای هر عضو یک کار
- `O(n log n)` → بسیار رایج در الگوریتم‌های مرتب‌سازی سریع
- `O(n²)` → معمولاً دو حلقهٔ وابسته به `n`
- `O(2^n)` → بررسی زیرمجموعه‌ها؛ برای `n` بزرگ خطرناک
- `O(n!)` → بررسی جایگشت‌ها؛ خیلی زود انفجاری می‌شود

<div class="warn">
Big-O «ثانیه» نیست و تعداد دقیق عملیات هم نیست. فقط رشد غالب را توصیف می‌کند.
مثلاً `3n + 10` و `100n + 2` هر دو از نظر رشد `O(n)` هستند.
</div>
</div>

````python
def linear_ops(n):
    ops = 0
    for _ in range(n):
        ops += 1
    return ops

def quadratic_ops(n):
    ops = 0
    for _ in range(n):
        for _ in range(n):
            ops += 1
    return ops

for n in [10, 100, 1000]:
    print(f"n={n:4} | O(n)≈{linear_ops(n):8} | O(n²)≈{quadratic_ops(n):10}")
````

<div class="dark"><h2 style="margin-top:0;">۷. جست‌وجوی دودویی (Binary Search)</h2></div>

### ایدهٔ کلیدی: «هر بار نصف گزینه‌ها را حذف کن»

اگر لیست **مرتب** باشد:

```text
[2, 5, 8, 11, 14, 17, 21, 30]
```

و دنبال `17` باشیم:

1. وسط را می‌بینیم.
2. اگر هدف بزرگ‌تر از وسط است، نیمهٔ چپ را کامل کنار می‌گذاریم.
3. دوباره وسط نیمهٔ باقی‌مانده را می‌بینیم.
4. ادامه می‌دهیم.

چون هر بار مسئله تقریباً نصف می‌شود، تعداد گام‌ها حدود `log₂(n)` است.

<div class="warn">
<b>شرط اصلی:</b> Binary Search روی دادهٔ نامرتب به این شکل معنی ندارد.
مرتب بودن همان اطلاعاتی است که اجازه می‌دهد یک نیمه را با اطمینان حذف کنیم.
</div>
</div>

````python
def binary_search_trace(a, target):
    left, right = 0, len(a) - 1
    step = 0

    while left <= right:
        step += 1
        mid = (left + right) // 2
        print(f"گام {step}: بازه=[{left},{right}]، mid={mid}، مقدار={a[mid]}")

        if a[mid] == target:
            return mid
        elif a[mid] < target:
            left = mid + 1
        else:
            right = mid - 1

    return -1

arr = [2, 5, 8, 11, 14, 17, 21, 30]
print("نتیجه:", binary_search_trace(arr, 17))
````

مرتب‌سازی فقط برای «قشنگ شدن خروجی» نیست. مرتب کردن می‌تواند ساختار پنهان مسئله را آشکار کند.

مثلاً بعد از مرتب‌سازی:

- کمینه و بیشینه کنار دو سر هستند.
- اعضای نزدیک کنار هم قرار می‌گیرند.
- Binary Search ممکن می‌شود.
- پیدا کردن تکراری‌ها آسان‌تر می‌شود.
- خیلی از روش‌های greedy و two pointers ممکن می‌شوند.

در Lecture 3 الگوریتم‌های Selection Sort، Bubble Sort و Merge Sort به‌عنوان نمونه‌های اصلی مطرح می‌شوند.
ما اینجا دو مورد اول را فقط برای فهم سازوکار بررسی می‌کنیم؛ برای کار واقعی Python معمولاً از `sorted` استفاده می‌کنیم.
</div>

````python
a = [5, 2, 7, 4, 1, 6, 3, 0]

print("اصل:", a)
print("مرتب:", sorted(a))
print("اصل هنوز تغییر نکرده:", a)

a.sort()
print("بعد از a.sort():", a)
````

<div class="rtl-box">
### ۸.۱ Selection Sort — هر بار کوچک‌ترین را انتخاب کن

ایده:

1. در قسمت مرتب‌نشده کوچک‌ترین عضو را پیدا کن.
2. آن را با خانهٔ فعلی جابه‌جا کن.
3. یک خانه جلو برو.

برای `n` عضو، تقریباً باید مجموع زیر را مقایسه کنیم:

```text
n + (n-1) + (n-2) + ... + 1
```

که از نظر رشد `O(n²)` است.
</div>

````python
def selection_sort_trace(a):
    a = a[:]  # کپی برای خراب نکردن ورودی
    n = len(a)

    for i in range(n):
        min_index = i

        for j in range(i + 1, n):
            if a[j] < a[min_index]:
                min_index = j

        a[i], a[min_index] = a[min_index], a[i]
        print(f"بعد از مرحلهٔ {i+1}: {a}")

    return a

selection_sort_trace([5, 2, 7, 4, 1])
````

<div class="rtl-box">
### ۸.۲ Bubble Sort — همسایه‌های نامرتب را جابه‌جا کن

در Bubble Sort بارها از روی لیست رد می‌شویم و هر دو همسایهٔ نامرتب را جابه‌جا می‌کنیم.

این الگوریتم برای آموزش مفهوم خوب است، اما برای مسابقه معمولاً انتخاب مناسبی نیست.
</div>

````python
def bubble_sort_trace(a):
    a = a[:]
    n = len(a)

    for pass_no in range(n - 1):
        changed = False
        for i in range(n - 1 - pass_no):
            if a[i] > a[i + 1]:
                a[i], a[i + 1] = a[i + 1], a[i]
                changed = True

        print(f"پاس {pass_no+1}: {a}")

        if not changed:
            break

    return a

bubble_sort_trace([5, 2, 7, 4, 1])
````

### ۸.۳ Merge Sort — فقط یک پیش‌نمایش برای جلسهٔ سوم

3. دو نصف مرتب را ادغام کن.

در Lecture 3 این الگوریتم با زمان `O(n log n)` مقایسه می‌شود.

فعلاً لازم نیست پیاده‌سازی‌اش را حفظ کنی. فقط این تصویر را به خاطر بسپار:

```text

دو مسئلهٔ کوچک‌تر
   ↓ حل کن

جواب نهایی
```

در جلسهٔ سوم آن را کامل‌تر خواهیم دید.

<div class="dark"><h2 style="margin-top:0;">۹. `set` و `dict` — گاهی «ساختمان داده» خودِ بهینه‌سازی است</h2></div>

فرض کن می‌خواهیم بدانیم آیا در لیست تکراری وجود دارد.

### راه ساده با لیست

هر عضو را با اعضای قبلی مقایسه کنیم → ممکن است `O(n²)` شود.

### راه بهتر با `set`

یک مجموعه از اعضای دیده‌شده نگه می‌داریم:

- اگر عضو قبلاً در `seen` بود → تکراری پیدا شده
- وگرنه آن را اضافه می‌کنیم

در پیاده‌سازی معمول Python، جست‌وجوی `set` و `dict` به‌طور متوسط بسیار سریع است و اغلب به‌صورت `O(1)` میانگین مدل می‌شود.
</div>

````python
def has_duplicate(a):
    seen = set()

    for x in a:
        if x in seen:
            return True
        seen.add(x)

    return False

print(has_duplicate([4, 7, 2, 9]))
print(has_duplicate([4, 7, 2, 7]))
````

<div class="rtl-box">
### ۹.۱ `dict` برای شمارش فراوانی

یکی از پرکاربردترین الگوهای مسابقه‌ای:

> «هر مقدار چند بار آمده است؟»
</div>

````python
def frequency_table(a):
    freq = {}

    for x in a:
        if x not in freq:
            freq[x] = 0
        freq[x] += 1

    return freq

print(frequency_table([2, 5, 2, 7, 5, 2]))
````

<div class="dark"><h2 style="margin-top:0;">۱۰. پشته (Stack) و صف (Queue)</h2></div>

در Lecture 5 دو ساختمان دادهٔ انتزاعی بسیار مهم معرفی می‌شوند:

### Stack — پشته

اصل **LIFO**:

> Last In, First Out  
> آخرین چیزی که وارد می‌شود، اولین چیزی است که خارج می‌شود.

مثال: دستهٔ بشقاب‌ها.

عملیات اصلی:

- `push` → افزودن
- `pop` → برداشتن از بالا

### Queue — صف

اصل **FIFO**:

> First In, First Out  
> اولین چیزی که وارد می‌شود، اولین چیزی است که خارج می‌شود.

مثال: صف بانک.

عملیات اصلی:

- `enqueue` → ورود به انتهای صف
- `dequeue` → خروج از ابتدای صف
</div>

````python
# Stack با list
stack = []

stack.append("A")
stack.append("B")
stack.append("C")

print("stack:", stack)
print("pop:", stack.pop())
print("بعد:", stack)
````

````python
from collections import deque

# Queue با deque
queue = deque()

queue.append("A")
queue.append("B")
queue.append("C")

print("queue:", queue)
print("dequeue:", queue.popleft())
print("بعد:", queue)
````

<div class="rtl-box">
### ۱۰.۱ چرا این دو برای جلسهٔ سوم مهم‌اند؟

- **DFS** را می‌توان با فکرِ پشته فهمید.
- **BFS** را با صف اجرا می‌کنیم.
- بررسی پرانتزها معمولاً با پشته انجام می‌شود.
- شبیه‌سازی صف‌ها، نوبت‌ها و فرآیندها با Queue طبیعی است.

### مثال: اعتبار پرانتزها

رشتهٔ زیر معتبر است:

```text
(()())
```

ولی این یکی نیست:

```text
())(
```

ایده: هر `(` را داخل پشته بگذار؛ برای هر `)` باید یک `(` قبلی برای جفت شدن وجود داشته باشد.
</div>

````python
def valid_parentheses(s):
    stack = []

    for ch in s:
        if ch == "(":
            stack.append(ch)
        elif ch == ")":
            if not stack:
                return False
            stack.pop()

    return len(stack) == 0

tests = ["(()())", "())(", "((()))", "(()"]
for t in tests:
    print(t, "->", valid_parentheses(t))
````

<div class="dark"><h2 style="margin-top:0;">۱۱. درخت و گراف — فقط مدل ذهنی</h2></div>

### گراف (Graph)

گراف از دو چیز ساخته می‌شود:

- **رأس (Vertex / Node)** → چیزها
- **یال (Edge)** → رابطه‌ها

مثال:

- شهرها = رأس، جاده‌ها = یال
- دانش‌آموزان = رأس، دوستی = یال
- صفحات وب = رأس، لینک = یال
- حالت‌های یک بازی = رأس، حرکت مجاز = یال

### درخت (Tree)

درخت نوع خاصی از گراف است که ساختار شاخه‌ای دارد و دور ندارد.

در Lecture 5، درخت جست‌وجوی دودویی، linked list و hash table به‌عنوان نمونه‌های ساختمان داده معرفی می‌شوند.
فعلاً برای ما مهم‌ترین پل به جلسهٔ سوم، خود مفهوم **گراف** و شیوهٔ نگهداری همسایه‌هاست.
</div>

<div class="rtl-box">
### ۱۱.۱ نمایش گراف با لیست مجاورت

فرض کن یال‌ها این‌ها هستند:

```text
0 -- 1
0 -- 2
1 -- 3

```

برای هر رأس، فهرست همسایه‌ها را نگه می‌داریم:

```python
graph = {
    0: [1, 2],
    1: [0, 3],
    2: [0, 3],
    3: [1, 2, 4],
    4: [3]

```

این نمایش را **Adjacency List** می‌نامیم.

<div class="rtl-box">
### ۱۱.۲ واژه‌هایی که باید بشناسی

- **همسایه**: رأسی که با یک یال مستقیم وصل است.
- **مسیر**: دنباله‌ای از رأس‌ها که با یال به هم وصل‌اند.
- **درجه**: تعداد یال‌های متصل به یک رأس (در گراف بدون جهت).
- **مؤلفهٔ همبند**: گروهی از رأس‌ها که بینشان مسیر وجود دارد.
- **visited**: مجموعهٔ رأس‌هایی که قبلاً دیده‌ایم.

<div class="exam">
واژهٔ `visited` را جدی بگیر. تقریباً هر پیمایش گراف به راهی نیاز دارد که دوباره در حلقه‌ها گیر نکند.
</div>

<div class="dark"><h2 style="margin-top:0;">۱۲. BFS و DFS — دو راه برای پیمایش گراف</h2></div>

### BFS — جست‌وجوی سطح‌به‌سطح

مثل موج از نقطهٔ شروع پخش می‌شود:

```text
فاصله 0: شروع
فاصله 1: همسایه‌های شروع
فاصله 2: همسایه‌های جدید آن‌ها
...
```

برای همین **صف (Queue)** ابزار طبیعی BFS است.

### DFS — تا جایی که می‌شود عمیق برو

یک مسیر را ادامه می‌دهد؛ وقتی دیگر نمی‌شود، برمی‌گردد.

DFS را می‌توان:

- با **پشته**
- یا با **بازگشت (recursion)**

نوشت.

در این فایل فقط نسخه‌های ساده را می‌بینیم؛ جلسهٔ سوم کاربردهای جدی‌تر را پوشش می‌دهد.
</div>

````python
from collections import deque

def bfs_order(graph, start):
    q = deque([start])
    visited = {start}
    order = []

    while q:
        u = q.popleft()
        order.append(u)

        for v in graph[u]:
            if v not in visited:
                visited.add(v)
                q.append(v)

    return order

print("BFS:", bfs_order(graph, 0))
````

````python
def dfs_order(graph, start):
    visited = set()
    order = []

    def dfs(u):
        visited.add(u)
        order.append(u)

        for v in graph[u]:
            if v not in visited:
                dfs(v)

    dfs(start)
    return order

print("DFS:", dfs_order(graph, 0))
````

<div class="rtl-box">
### ۱۲.۱ مهم: ترتیب BFS/DFS همیشه یکتا نیست

اگر ترتیب همسایه‌ها را عوض کنیم، ترتیب پیمایش هم می‌تواند عوض شود.

پس معمولاً سؤال اصلی این نیست که «دقیقاً چه ترتیبی؟»، بلکه این است:

- آیا به همهٔ رأس‌های قابل دسترس می‌رسیم؟
- فاصلهٔ کمینه چیست؟
- چند مؤلفه داریم؟
- آیا مسیر وجود دارد؟
</div>

<div class="rtl-box">
### ۱۲.۲ BFS و کوتاه‌ترین مسیر در گراف بدون وزن

اگر همهٔ حرکت‌ها هزینهٔ یکسان داشته باشند، BFS اولین بار که به یک رأس می‌رسد، کمترین تعداد یال از شروع را پیدا کرده است.
</div>

````python
def bfs_distance(graph, start):
    q = deque([start])
    dist = {start: 0}

    while q:
        u = q.popleft()

        for v in graph[u]:
            if v not in dist:
                dist[v] = dist[u] + 1
                q.append(v)

    return dist

print(bfs_distance(graph, 0))
````

<div class="dark"><h2 style="margin-top:0;">۱۳. فضای حالت — مهم‌ترین پل به مسائل المپیاد</h2></div>

گاهی سؤال اصلاً دربارهٔ «شهر و جاده» نیست، اما باز هم گراف است.

مثلاً:

> از عدد 10 شروع می‌کنیم. در هر حرکت می‌توانیم 1 واحد کم کنیم یا اگر زوج بود، بر 2 تقسیم کنیم. کمترین تعداد حرکت برای رسیدن به 1 چقدر است؟

می‌توانیم بگوییم:

- هر **عدد** یک حالت / رأس است.
- هر **عمل مجاز** یک یال است.
- دنبال کوتاه‌ترین مسیر از `10` به `1` هستیم.

این یعنی یک مسئلهٔ عددی را به **گراف فضای حالت** تبدیل کرده‌ایم.
</div>

````python
def next_states(x):
    result = []

    if x > 1:
        result.append(x - 1)

    if x % 2 == 0:
        result.append(x // 2)

    return result

for x in [10, 9, 8, 7]:
    print(x, "->", next_states(x))
````

````python
def min_moves_to_one(start):
    q = deque([(start, 0)])
    visited = {start}

    while q:
        x, d = q.popleft()

        if x == 1:
            return d

        for y in next_states(x):
            if y not in visited:
                visited.add(y)
                q.append((y, d + 1))

print("کمترین حرکت از 10 به 1:", min_moves_to_one(10))
````

````python
# تمرین ۱۳: کمترین حرکت از 1 تا 23 با +1 و ×2
def min_moves_up(target):
    q = deque([(1, 0)])
    visited = {1}

    while q:
        x, d = q.popleft()

        if x == target:
            return d

        for y in [x + 1, x * 2]:
            if y <= target and y not in visited:
                visited.add(y)
                q.append((y, d + 1))

print("کمترین تعداد حرکت:", min_moves_up(23))
````
